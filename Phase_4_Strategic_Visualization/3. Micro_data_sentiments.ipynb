# STAGE 1: Data Cleaning & Preparation

```python
def cleanMicroDataset(df):
    """
    1. Select relevant columns
    2. Create a new column 'reviewText', take the value of 'textTranslated'.
    3. If 'textTranslated' is null, take the value of 'text'.
    4. Remove null row in 'reviewText'
    5. Drop 'textTranslated' and 'text' columns
    """
    df = df[['title', 'placeId','publishedAtDate', 'stars', 'text', 'textTranslated']]
    df['reviewText'] = df['textTranslated'].fillna(df['text'])
    df = df[~df["reviewText"].isna()]
    df = df.drop(columns =['text', 'textTranslated'])

    return df
```

```python
def segment_review(text):
    """
    Splits review into clauses using punctuation and conjunctions.
    (Simple Clause Segmentation)
    """
    if pd.isna(text):
        return []

    # Split by common delimiters: . , ! ? and 'but'
    # We use regex to keep the delimiters or just split clean
    # Here we split by punctuation or 'but'/'and'
    segments = re.split(r'[.,!?]|\bbut\b|\band\b', str(text).lower())

    # Clean up whitespace and ignore empty segments
    return [seg.strip() for seg in segments if len(seg.strip()) > 3]
```

```python
def identify_aspects_fuzzy(segment):
    """
    Maps a text segment to aspects using Exact Match AND Fuzzy Match.
    Includes 'Emoji Protection' to stop warnings.
    """
    found_aspects = set()
    words = segment.split()

    for word in words:
        # CLEANUP: Remove punctuation
        clean_word = word.strip(".,!?").lower()

        if len(clean_word) < 3:
            continue

        # Skip emoji
        # If the word has no letters (a-z), it is likely an emoji or number. Skip it.
        if not re.search('[a-zA-Z]', clean_word):
            continue

        # --- PASS 1: EXACT MATCH  ---
        if clean_word in KEYWORD_TO_ASPECT:
            found_aspects.add(KEYWORD_TO_ASPECT[clean_word])
            continue

        # --- PASS 2: FUZZY MATCH  ---
        if len(clean_word) > 4:
            match, score = process.extractOne(clean_word, unique_keywords)

            if score >= 85:
                found_aspects.add(KEYWORD_TO_ASPECT[match])

    if not found_aspects:
        return ['GENERAL']

    return list(found_aspects)
```

```python
print("Running Aspect Extraction with Fuzzy Matching on Micro Dataset...")

results = []

for index, row in df_micro.iterrows():
    segments = segment_review(row['reviewText'])

    for seg in segments:
        aspects = identify_aspects_fuzzy(seg)

        for aspect in aspects:
            results.append({
                'Original_Review_ID': index,
                'title': row['title'],
                'placeId': row['placeId'],
                'publishedAtDate': row['publishedAtDate'],
                'stars': row['stars'],
                'Full_Review': row['reviewText'],
                'Segment': seg,
                'Aspect': aspect,
            })

df_aspects = pd.DataFrame(results)
df_aspects_clean = df_aspects[df_aspects['Aspect'] != 'GENERAL'].reset_index(drop=True)

print(f"Fuzzy Extraction Complete! Total segments: {len(df_aspects_clean)}")
print(df_aspects_clean['Aspect'].value_counts())
```

# STAGE 3: Sentiment Classfication

```python
import torch
import torch.nn as nn
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from torch.utils.data import Dataset, DataLoader
from transformers import DataCollatorWithPadding

# ── Constants (must match training config) ─────────────────────────────────
MODEL_NAME    = "xlm-roberta-base"
MODEL_PATH    = r"/content/drive/MyDrive/Aspect-Based-Sentiment-Analysis/Modelling/models/xlm_roberta_absa_best_after_filtering.pt"
MAX_SEQ_LEN   = 128
BATCH_SIZE    = 32
ID2LABEL      = {0: "negative", 1: "positive"}
CLASS_WEIGHTS = [6.0, 1.0]   # Same as training

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {DEVICE}")


# ── Replicate model architecture (must match training exactly) ──────────────
class ABSASentimentClassifier(nn.Module):
    def __init__(self, model_name, num_labels, class_weights):
        super().__init__()
        self.backbone = AutoModelForSequenceClassification.from_pretrained(
            model_name, num_labels=num_labels
        )
        self.register_buffer(
            "class_weights",
            torch.tensor(class_weights, dtype=torch.float),
        )

    def forward(self, input_ids, attention_mask, labels=None):
        outputs = self.backbone(input_ids=input_ids, attention_mask=attention_mask)
        return {"logits": outputs.logits}


# ── Load tokenizer & model ──────────────────────────────────────────────────
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = ABSASentimentClassifier(MODEL_NAME, num_labels=2, class_weights=CLASS_WEIGHTS)
# strict=False: ignores 'loss_fn.class_weight' saved from FocalLoss during training
# — not needed for inference, only used during the training forward pass
model.load_state_dict(torch.load(MODEL_PATH, map_location=DEVICE), strict=False)
model.to(DEVICE)
model.eval()

print("Model loaded successfully.")
```

```python
class SegmentDataset(Dataset):
    """Minimal dataset for inference — no labels required."""
    def __init__(self, texts, tokenizer, max_length):
        self.encodings = tokenizer(
            texts,
            max_length=max_length,
            truncation=True,
            padding=False,
            return_tensors=None,
        )

    def __len__(self):
        return len(self.encodings["input_ids"])

    def __getitem__(self, idx):
        return {
            "input_ids":      self.encodings["input_ids"][idx],
            "attention_mask": self.encodings["attention_mask"][idx],
        }


# ── Run inference on all segments ──────────────────────────────────────────
segments = df_aspects_clean["Segment"].tolist()

dataset     = SegmentDataset(segments, tokenizer, MAX_SEQ_LEN)
collator    = DataCollatorWithPadding(tokenizer=tokenizer)
loader      = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=False, collate_fn=collator)

all_preds = []

print(f"Running inference on {len(segments):,} segments...")

with torch.no_grad():
    for batch in loader:
        input_ids      = batch["input_ids"].to(DEVICE)
        attention_mask = batch["attention_mask"].to(DEVICE)
        outputs        = model(input_ids=input_ids, attention_mask=attention_mask)
        preds          = torch.argmax(outputs["logits"], dim=-1)
        all_preds.extend(preds.cpu().numpy().tolist())

df_aspects_clean["predicted_sentiment"] = [ID2LABEL[p] for p in all_preds]

print("Inference complete!")
print(df_aspects_clean["predicted_sentiment"].value_counts())
```

```python
print("="*70)
print("SEGMENT-LEVEL SENTIMENT CONSISTENCY CHECK")
print("="*70)

# 1. Initialize VADER
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
vader = SentimentIntensityAnalyzer()

print("\nCalculating VADER scores for each segment...")

# 2. Calculate VADER score for each segment
df_aspects_clean['segment_vader_score'] = df_aspects_clean['Segment'].apply(
    lambda x: vader.polarity_scores(str(x))['compound'] if pd.notna(x) and str(x).strip() != "" else 0.0
)

print(f"✓ VADER scores computed for {len(df_aspects_clean):,} segments")

# 3. Define consistency check function
def check_segment_consistency(row):
    """
    Check if segment sentiment matches inherited weak label.
    Returns True if consistent, False if inconsistent.

    Following Chen et al. (2025) and Sharifpour & Lahmiri (2023):
    - Positive label: Keep unless STRONGLY negative (score < -0.3)
    - Negative label: Keep unless STRONGLY positive (score > 0.3)
    """
    score = row['segment_vader_score']
    label = row['predicted_sentiment']

    if label == 'positive':
        return score > -0.3
    elif label == 'negative':
        return score < 0.3
    else:
        return True

# 4. Apply consistency check
df_aspects_clean['is_segment_consistent'] = df_aspects_clean.apply(
    check_segment_consistency, axis=1
)

# 7. Filter to keep only consistent segments
df_aspects_final = df_aspects_clean[
    df_aspects_clean['is_segment_consistent']
].copy()

# 8. Final statistics
print(f"\n{'='*70}")
print("FINAL DATASET SUMMARY")
print(f"{'='*70}")
print(f"Before filtering: {len(df_aspects_clean):,} segments")
print(f"After filtering:  {len(df_aspects_final):,} segments")
print(f"Removed:          {len(df_aspects_clean) - len(df_aspects_final):,} segments")

print(f"\nLabel distribution (BEFORE filtering):")
print(df_aspects_clean['predicted_sentiment'].value_counts())

print(f"\nLabel distribution (AFTER filtering):")
print(df_aspects_final['predicted_sentiment'].value_counts())

print(f"\nAspect distribution (AFTER filtering):")
# Explode aspect labels to count individual aspects
aspect_counts = df_aspects_final.explode('predicted_sentiment')['predicted_sentiment'].value_counts()
print(aspect_counts)
```
